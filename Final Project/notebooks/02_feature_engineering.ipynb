```python
import pandas as pd

DATA_PATH = "../data/raw/"

orders = pd.read_csv(DATA_PATH + "orders.csv")
order_products_prior = pd.read_csv(DATA_PATH + "order_products__prior.csv")
products = pd.read_csv(DATA_PATH + "products.csv")
```

```python
user_features = orders.groupby("user_id").agg({
    "order_number": "max"
}).rename(columns={
    "order_number": "total_orders"
})

user_features.head()
```

```python
avg_days = orders.groupby("user_id")["days_since_prior_order"].mean()

user_features["avg_days_between_orders"] = avg_days
```

```python
user_order = order_products_prior.merge(
    orders[["order_id", "user_id"]],
    on="order_id",
    how="left"
)

reorder_ratio = user_order.groupby("user_id")["reordered"].mean()

user_features["reorder_ratio"] = reorder_ratio
```

```python
user_features.head()
```

```python
user_features.to_csv(
    "../data/processed/user_features.csv"
)
```
